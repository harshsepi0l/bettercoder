# Forest fire (small)


```
%matplotlib inline
EMPTY, TREE, FIRE = 0, 1, 2
```

```
def spread(grid):
    new = [row[:] for row in grid]
    for r in range(len(grid)):
        for c in range(len(grid[r])):
            if grid[r][c] == FIRE:
                new[r][c] = EMPTY
                if c + 1 < len(grid[r]) and grid[r][c + 1] == TREE:
                    new[r][c + 1] = FIRE
    return new
```

Start a fire on the left and let it spread twice.


```
forest = [[FIRE, TREE, TREE], [TREE, TREE, EMPTY]]
for step in range(2):
    forest = spread(forest)
print(forest)
```
